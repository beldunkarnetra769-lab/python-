```python
# Bank loan approval → Country → Employment → Salary → Credit score
country = input("Enter your country:")
employment = input("Are you employed? (yes/no):")
salary = float(input("Enter your monthly salary:"))
credit_score = int(input("Enter your credit score:"))

if country.lower() == "india":
    if employment.lower() == "yes":
        if salary >= 30000:
            if credit_score >= 700:
                print("Loan Approved")
            else:
                print("Loan Rejected: Credit score is too low")
        else:
            print("Loan Rejected: Salary is too low")
    else:
        print("Loan Rejected: You are not employed")
else:
    print("Loan Rejected: Country not eligible")
```

```python
#College admission → State → Marks → Age → Qualification	
state = input("Enter your state: ")
marks = int(input("Enter your marks: "))
age = int(input("Enter your age: "))
qualification = input("Enter your qualification: ")

if state.lower() == "karnataka":
    \
    if marks >= 50:
        if age >= 18:
            if qualification.lower() == "bca":
                print("Admission Eligible")
            else:
                print("Admission Rejected: Qualification is not BCA")
        else:
            print("Admission Rejected: Age is below 18")
    else:
        print("Admission Rejected: Marks are below 50")
else:
    print("Admission Rejected: State is not eligible")
```

```python
# Online shopping → Login → Product availability → Payment → Order confirmation
login = input("Are you logged in? (yes/no): ")
product = input("Is the product available? (yes/no): ")
payment = input("Is payment successful? (yes/no): ")

if login.lower() == "yes":
    if product.lower() == "yes":
        if payment.lower() == "yes":
            print("Order Confirmed")
        else:
            print("Order Failed: Payment unsuccessful")
    else:
        print("Order Failed: Product not available")
else:
    print("Order Failed: Please login first")
```

```python
# ATM withdrawal → Card valid → PIN correct → Balance available → Withdraw
card = input("Is the card valid? (yes/no): ")
pin = input("Is the PIN correct? (yes/no): ")
balance = float(input("Enter your balance: "))
amount = float(input("Enter withdrawal amount: "))

if card.lower() == "yes":
    if pin.lower() == "yes":
        if balance >= amount:
            print("Withdrawal Successful")
        else:
            print("Withdrawal Failed: Insufficient balance")
    else:
        print("Withdrawal Failed: Incorrect PIN")
else:
    print("Withdrawal Failed: Invalid card")
```

```python
#Insurance eligibility → Country → Age → Income → Medical/other criteria	
country = input("Enter your country: ")
age = int(input("Enter your age: "))
income = float(input("Enter your monthly income: "))
medical = input("Medical criteria satisfied? (yes/no): ")

if country.lower() == "india":
    if age >= 18:
        if income >= 20000:
            if medical.lower() == "yes":
                print("Insurance Eligible")
            else:
                print("Insurance Rejected: Medical criteria not satisfied")
        else:
            print("Insurance Rejected: Income is too low")
    else:
        print("Insurance Rejected: Age is below 18")
else:
    print("Insurance Rejected: Country not eligible")
```

```python
#Employee promotion → Department → Experience → Performance → Salary criteria	
department = input("Enter your department: ")
experience = int(input("Enter years of experience: "))
performance = input("Is performance good? (yes/no): ")
salary = float(input("Enter your salary: "))

if department.lower() == "IT":
    if experience >= 2:
        if performance.lower() == "yes":
            if salary >= 20000:
                print("Eligible for Promotion")
            else:
                print("Promotion Rejected: Salary criteria not satisfied")
        else:
            print("Promotion Rejected: Performance is not good")
    else:
        print("Promotion Rejected: Experience is less than 2 years")
else:
    print("Promotion Rejected: Department is not eligible")
```
