```python
#Write a function `is_palindrome(s)
def plaindroms(s):
    s=s.lower()
    return s==s[::-1]
```

```python
#Write a function `count_vowels(s)
def count_vowels(s):
    count = 0

    for ch in s:
        if ch.lower() in "aeiou":
            count = count + 1

    return count
```

```python
count_vowels("hello world")
```

```python
# Write a function `factorial(n)
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result
```

```python
factorial(5)
```

```python
#Write a function `is_prime(n)

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
```

```python
is_prime(3)
```

```python
# Write a function `average(numbers)
def average(numbers):
    total = 0

    for num in numbers:
        total = total + num

    return total / len(numbers)
```

```python
average([2, 4, 6])
```

```python
#Write a function `find_max(numbers)
def find_max(numbers):
    maximum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum
```

```python
find_max((4,6,1,8,9,33))
```

```python
# Write a function `remove_duplicates(lst)
def remove_duplicates(lst):
    result = []

    for item in lst:
        if item not in result:
            result.append(item)

    return result
```

```python
remove_duplicates([2, 1,1,3, 4, 2])
```

```python
#Write a function `celsius_to_fahrenheit(c)
def celsius_to_fahrenheit(c):
    f = c * 9/5 + 32
    return f


```

```python
#Write a function `count_occurrences(lst, item)
def count_occurrences(lst, item):
    count = 0
    
    for i in lst:
        if i == item:
            count = count + 1
            
    return count
```

```python
count_occurrences([1, 2, 2, 3, 2], 2)
```

```python
#Question 15: FizzBuzz
def fizzbuzz(n):
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return str(n)
```

```python
fizzbuzz(4)
```
