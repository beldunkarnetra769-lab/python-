## loops


```python
#while 
i=0
while i<5:
    i+=1
    print(i)
```

```python
# print even numbers  1 to 15  using while loops
i=0
while i<15:
    print(i)
    i+=2
```

```python
# print  odd numbers
i=1
while i<15:
    print(i)
    i+=2
```

```python
# continue(skip the condition)
i=0
while i<5:
    i=i+1
    if i==3:
        continue
    print(i)
```

```python
# break (exit the loop)
i=0
while i<5:
    i=i+1
    if i==3:
        break
    print(i)
```

```python
# for loop
num=[1,2,3,4,5,12,21,35,78,10,14,15]
for i in num:
    if i%2==0:
        print(i)
    
```

```python
# ood number
num=[1,2,3,4,5,12,21,35,78,10,14,15]
for i in num:
    if i%2==1:
        print(i)
```

```python
# odd or even numbers
num=[1,2,3,4,5,12,21,35,78,10,14,15]
for i in num:
    if i%2==0:
        print(i,"even number")
    else:
        print(i,"odd number")
```
