```python
#nested if 
age=int(input("enter the age"))
if age<18 and age!=0:
    print("minner")
elif age<50 and age!=0:
    print(" adult")
elif age<=100 and age!=0:
    print("senior citizen")
else:
    print("invalid")
```

```python
# grading system
marks=int(input("enter your marks"))

if marks<=35 and marks<0:
    print("fail")
elif marks<=50 and marks!=0:
    print("B")
elif marks<=70 and marks!=0:
     print("B+")
elif marks<=70 and marks!=0:
    print("A")
elif marks<=100 and marks!=0:
    print("A+")
else:
    print("invalid")
```

```python
# rating  systrem
rating=int(input("enter your rating"))
if rating==1:
    print("poor")
elif rating==2:
    print("average")
elif rating==3:
    print("good")
elif rating==4:
    print("very good")
elif rating==5:
    print("Awesome")
else:
    print("invalid")
```

```python
country = input("Enter your country:")
employment = input("Are you employed? (yes/no):")
salary = float(input("Enter your monthly salary:"))
credit_score = int(input("Enter your credit score:"))

if country.lower() == "india":
    if employment.lower() == "yes":
        if salary >= 30000:
            if credit_score >= 700:
                print("Loan Approved")
            else:
                print("Loan Rejected: Credit score is too low")
        else:
            print("Loan Rejected: Salary is too low")
    else:
        print("Loan Rejected: You are not employed")
else:
    print("Loan Rejected: Country not eligible")
```
