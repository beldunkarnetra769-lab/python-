### TYPE CASTING

```python
name = "Bhavana"
age = 22
percentage = 85.5
is_pass = True
```

```python
# Integer to Float
age = 22
float(age)
```

```python
# Float to Integer
percentage = 85.5
int(percentage)
```

```python
# Integer to String
age = 22
str(age)
```

```python
# String to Integer
marks = "90"
int(marks)
```

```python
# Boolean to Integer
is_pass = True
int(is_pass)
```

```python
# Integer to Boolean
number = 1
bool(number)
```

```python
# String with letters cannot convert to Integer
name = "Bhavana"
int(name)
```

```python
# String with letters cannot convert to Float
name = "Bhavana"
float(name)
```

```python
# Decimal String cannot directly convert to Integer
number = "85.5"
int(number)
```

```python
# None cannot convert to Integer
value = None
int(value)
```

```python
# None cannot convert to Float
value = None
float(value)
```
